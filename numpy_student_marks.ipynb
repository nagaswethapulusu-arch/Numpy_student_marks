```python
import numpy as np

students = ["Asha", "Ravi", "Meena", "Kiran", "Swetha", "Harsh", "Rupesh", "Akhil", "Divya", "Teja"]
marks = np.array([78, 85, 62, 90, 71, 89, 70, 65, 92, 55])

print("Marks array:", marks)
print("Type:", type(marks))
print("Shape:", marks.shape)
print("Data type:", marks.dtype)
```

```python
mean = np.mean(marks)
median = np.median(marks)
maximum = np.max(marks)
minimum = np.min(marks)
std_dev = np.std(marks)
```

```python
print(f"Mean: {mean:.2f}")
print(f"Median: {median:.2f}")
print(f"Maximum: {maximum}")
print(f"Minimum: {minimum}")
print(f"Standard deviation: {std_dev:.2f}")
```

```python
print("Topper:", students[np.argmax(marks)], "-", maximum)
print("Lowest:", students[np.argmin(marks)], "-", minimum)

above_mean = marks > mean
print("Marks above mean:", marks[above_mean])
print("Students above mean:", [s for s, flag in zip(students, above_mean) if flag])

print(f"Population std (ddof=0): {np.std(marks):.2f}")
print(f"Sample std (ddof=1): {np.std(marks, ddof=1):.2f}")
```
